# Mã hóa các câu trả lời

```python
import pandas as pd

# =========================
# 1. Standardize column names
# =========================
df = df.rename(columns={
    "Bạn hiện đang là học sinh lớp mấy?": "doi_tuong",
    "Bạn đến từ miền nào? (Bao gồm các khu vực hải đảo tương ứng)": "dia_diem",
    "Bạn giỏi những môn nào trong những môn bạn dự tính thi THPTQG?": "mon_gioi",
    "Bạn không giỏi những môn nào trong những môn bạn dự tính thi THPTQG?": "mon_yeu",
    "Bạn dự đoán điểm tổ hợp của bạn sẽ trong khoảng nào?": "diem_du_doan",
    "Điểm trung bình trong lớp của bạn như thế nào ở học kỳ gần nhất?": "diem_gan_nhat",
    "Bạn dự định sẽ thi những tổ hợp nào? \n(Nếu chưa quyết định, em hãy điền tổ hợp mà em đang có xu hướng chọn cao nhất nhé.)": "to_hop_du_dinh",
    "Mục đích bạn thi THPTQG là gì?": "muc_dich",
    "Giả sử bạn vừa biết điểm thi tốt nghiệp THPTQG, bạn muốn biết điều gì về kết quả của mình?": "muc_tieu_chinh",
    "Ngoài những mong muốn trên, bạn còn khao khát điều gì khác khi cầm tờ kết quả trên tay? (Ví dụ: Muốn biết mình có thuộc Top 100 của tỉnh không, hay muốn so sánh điểm với bạn bè một cách ẩn danh...)  ": "muc_tieu_phu"
})



# =========================
# 2. Filter valid respondents
# =========================
valid_groups = [
    "Lớp 10",
    "Lớp 11",
    "Lớp 12",
    "Thí sinh tự do",
    "Cựu học sinh / sinh viên (Đã tốt nghiệp THPT)"
]

df = df[df["doi_tuong"].isin(valid_groups)]



# =========================
# 3. Grouping target population
# =========================
df["doi_tuong"] = df["doi_tuong"].replace({
    "Thí sinh tự do": "Đã tốt nghiệp THPT",
    "Cựu học sinh / sinh viên (Đã tốt nghiệp THPT)": "Đã tốt nghiệp THPT"
})


# =========================
# 4. Feature engineering (optional semantic cleaning)
# =========================
# (giữ raw text, chuẩn hóa câu trả lời)
df["dia_diem"] = df["dia_diem"].replace({
    "Miền Bắc (bao gồm Trung du và miền núi Bắc Bộ, và Đồng bằng sông Hồng)" : "Miền Bắc",
    "Miền Trung (bao gồm Bắc Trung Bộ, Duyên hải Nam Trung Bộ, và Tây Nguyên)" : "Miền Trung", 
    "Miền Nam (bao gồm Đông Nam Bộ và Đồng bằng sông Cửu Long)" : "Miền Nam"
}).astype("category")



# =========================
# 5. Drop non-informative respondents (optional rule-based filtering)
# =========================
df = df[~df["doi_tuong"].isin(["Cha mẹ", "Thầy cô"])]




# =========================
# Resulting dataset ready for analysis
# =========================
clean_df = df.copy()

clean_df.head(10)
```

```python
clean_df["valid"] = True
```

# Tiền xử lý dữ liệu

## Xử lý dữ liệu môn học

```python
valid_subjects = [
    "Toán (Đại số / Giải tích)",
    "Toán (Hình học)",
    "Ngữ văn",
    "Tiếng Anh",
    "Vật lý",
    "Hóa học",
    "Sinh học",
    "Lịch sử",
    "Địa lý",
    "GDKT&PL (Hoặc GDCD)",
    "Tin học",
    "Công nghệ công nghiệp",
    "Công nghệ nông nghiệp"
]


valid_subjects_set = set(valid_subjects)
```

```python
# B1: Tách chuỗi
def normalize_subject(x):
    if pd.isna(x):
        return []
    
    return [
        val.strip() for val in str(x).split(";") if val.strip()
    ]

clean_df["mon_gioi"] = clean_df["mon_gioi"].apply(normalize_subject)
clean_df["mon_yeu"] = clean_df["mon_yeu"].apply(normalize_subject)
```

```python
# B2: Kiểm tra tính hợp lệ

# 0. Khởi tạo tất cả đều là True
clean_df["valid"] = True

# 1. Kiểm tra môn lạ (Out of bounds)
# Nếu mon_gioi hoặc mon_yeu chứa bất kỳ phần tử nào không nằm trong valid_subjects_set
def has_invalid(subject_list):
    return any(s for s in subject_list if s not in valid_subjects_set)

invalid_gioi_mask = clean_df["mon_gioi"].apply(has_invalid)
invalid_yeu_mask = clean_df["mon_yeu"].apply(has_invalid)
```

## Kiểm tra trùng môn

```python
def has_conflict(row):
    return not set(row["mon_gioi"]).isdisjoint(set(row["mon_yeu"]))

conflict_mask = clean_df.apply(has_conflict, axis=1)


# --- GÁN TRỰC TIẾP ---
# Chỉ cần vi phạm 1 trong các mask trên, valid sẽ thành False
clean_df.loc[invalid_gioi_mask | invalid_yeu_mask | conflict_mask, "valid"] = False
```

## Tổ hợp

```python
bang_to_hop = pd.read_csv("input/bang_to_hop_mon.csv")

mapping_to_hop = {}
for _, row in bang_to_hop.iterrows():
    mapping_to_hop[row['Tổ hợp']] = {row['Môn 1'], row['Môn 2'], row['Môn 3']}
```

```python
# Quy các môn Toán về cùng 1 môn

subject_mapping = {
    "Toán (Đại số / Giải tích)": "Toán",
    "Toán (Hình học)": "Toán",
    "Ngữ văn": "Văn",
    "Tiếng Anh": "Anh",
    "Vật lý": "Lý",
    "Hóa học": "Hóa",
    "Sinh học": "Sinh",
    "Lịch sử": "Sử",
    "Địa lý": "Địa",
    "GDKT&PL (hoặc Giáo dục công dân)": "Giáo dục kinh tế và pháp luật",
    "Tin học": "Tin"
    # Hai môn Công nghệ giữ nguyên nên không cần map
}

def collapse_subjects(subject_list):
    # Chuyển đổi và dùng set() để loại bỏ trùng lặp 
    # (Tránh việc một dòng có cả Đại và Hình sẽ bị biến thành ['Toán', 'Toán'])
    collapsed = {subject_mapping.get(s, s) for s in subject_list}
    
    return list(collapsed)

# Áp dụng cho cả 2 cột
clean_df["mon_gioi"] = clean_df["mon_gioi"].apply(collapse_subjects)
clean_df["mon_yeu"] = clean_df["mon_yeu"].apply(collapse_subjects)
```

```python
import re

def extract_combinations(text):
    if pd.isna(text):
        return []
    
    # 1. Chuyển về chữ thường để dễ xử lý
    text = str(text).lower()
    
    # 2. Dùng Regex để tìm các mã khối (Ví dụ: a00, b01, d07...)
    # Pattern: 1 chữ cái + 2 hoặc 3 chữ số
    codes = re.findall(r'[a-z]\d{2,3}', text)
    
    # 3. Chuẩn hóa về chữ hoa
    valid_codes = [c.upper() for c in codes]
    
    # Nếu không tìm thấy mã khối nào, trả về ['INVALID'] để sau này lọc bỏ
    return valid_codes if valid_codes else ["INVALID"]


col_name = 'to_hop_du_dinh'
clean_df['to_hop_clean'] = clean_df[col_name].apply(extract_combinations)
```

```python
def check_target_logic(row):
    if not row["valid"]: 
        return False
    
    user_combos = row["to_hop_clean"]
    mon_gioi = set(row["mon_gioi"])
    if not mon_gioi:
        return True
    
    for combo in user_combos:
        if combo in mapping_to_hop:
            # Nếu mục tiêu là combo này, nhưng mon_gioi không chứa môn nào trong đó
            if not (mon_gioi & mapping_to_hop[combo]):
                return False
    return True

clean_df["valid"] = clean_df.apply(check_target_logic, axis=1)
```

# Xử lý mong muốn chính

```python
valid_targets = [
    "Vị thế tương đối", 
    "Xác suất \"an toàn\"", 
    "Thứ hạng ước tính", 
    "Phân tích phổ điểm", 
    "So sánh tổ hợp"
]

def process_multi_choice_target(text):
    if pd.isna(text): return []
    
    # Tách các lựa chọn dựa trên dấu phẩy
    choices = [c.strip() for c in str(text).split(';')]
    
    final_choices = []
    has_junk = False
    
    for c in choices:
        if c in valid_targets:
            final_choices.append(c)
        else:
            # Đây là phần "điền tay"
            # Nếu điền tay mà có nghĩa (độ dài > 5 và không phải ký tự rác)
            if len(c) > 5 and not any(char in c for char in [".", "x", "asdf"]):
                final_choices.append("Custom: " + c)
            else:
                has_junk = True # Dính bẫy điền bừa
                
    return final_choices, has_junk

# Áp dụng
res = clean_df.iloc[:, 9].apply(process_multi_choice_target)
clean_df['target_list'] = res.apply(lambda x: x[0])
clean_df['hit_target_trap'] = res.apply(lambda x: x[1])

# Nếu dính bẫy (điền tay rác rưởi) thì đánh dấu invalid
clean_df.loc[clean_df['hit_target_trap'], 'valid'] = False
```

# Xử lý mong muốn phụ

```python
def clean_khao_khat(text):
    if pd.isna(text):
        return None
    
    # Chuẩn hóa: viết thường, bỏ khoảng trắng thừa
    t = str(text).lower().strip(";")
    
    # 1. Danh sách các câu trả lời "vô thưởng vô phạt"
    blacklist = [
        'không', 'không có', 'không có mong muốn nào khác', 'không còn', 
        'ko', 'k', 'khong', 'khônng', 'khômg', 'none', 'no', 'k bít', 'ko biết', 
        'không biết', 'không ạ', 'em không muốn trả lời ạ...', 'chưa có',
        'chúc anh chị thi vào ngôi trường đại học mà mình mong muốn!!', # Câu chúc, không phải mong muốn bản thân
        'dạ em không có ý kiến..', 'không viết', 'ngủ', 'nghỉ hè'
    ]
    
    # 2. Lọc theo danh sách đen
    if t in blacklist:
        return None
    
    # 3. Lọc theo độ dài và ký tự rác (như "shbsbaah", "..", "^^")
    if len(t) < 3 or t.startswith('^^') or t.startswith('..'):
        return None
        
    # Nếu vượt qua các bộ lọc trên, trả về nội dung gốc (đã strip)
    return str(text).strip()

# Áp dụng
clean_df['khao_khat_clean'] = clean_df.iloc[:, 10].apply(clean_khao_khat)
```
